# Задание 2

```python
def staff_feedback(feedback):
    avg_rating = sum(person["rating"] for person in feedback) / len(feedback)

    rude_mentions = sum(
        comment.lower() == "rude"
        for person in feedback
        for comment in person["comments"]
    )

    best_staff = max(feedback, key=lambda person: person["rating"])["staff"]

    return {
        "avg_rating": round(avg_rating, 1),
        "rude_mentions": rude_mentions,
        "best_staff": best_staff,
    }


feedback = [
    {"staff": "Мария", "rating": 9, "comments": ["friendly", "helpful"]},
    {"staff": "Иван", "rating": 7, "comments": ["slow", "polite", "rude"]},
    {"staff": "Ольга", "rating": 10, "comments": ["amazing", "professional"]},
    {"staff": "Антон", "rating": 5, "comments": ["rude", "unfriendly"]},
]

print(staff_feedback(feedback))
```

# Задание 3

```python
def stay_duration_analysis(bookings):
    longest_stay = max(bookings, key=lambda booking: booking["nights"])

    nights_by_type = {}
    for booking in bookings:
        room_type = booking["room_type"]
        nights_by_type.setdefault(room_type, []).append(booking["nights"])

    average_by_room_type = {
        room_type: sum(nights) / len(nights)
        for room_type, nights in nights_by_type.items()
    }

    return {
        "longest_stay_guest": longest_stay["guest"],
        "longest_stay_nights": longest_stay["nights"],
        "average_nights_by_room_type": average_by_room_type,
    }


bookings = [
    {"guest": "Анна", "nights": 2, "room_type": "standard"},
    {"guest": "Иван", "nights": 10, "room_type": "suite"},
    {"guest": "Мария", "nights": 5, "room_type": "standard"},
    {"guest": "Павел", "nights": 1, "room_type": "economy"},
]

print(stay_duration_analysis(bookings))
```
